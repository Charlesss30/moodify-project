```python
# ==========================================
# CÀI ĐẶT CÁC THƯ VIỆN CẦN THIẾT
# !pip install transformers datasets pyvi scikit-learn pandas

import torch
import torch.nn as nn
import pandas as pd
import os
import json
import logging
import re
from pyvi import ViTokenizer
from torch.utils.data import Dataset, DataLoader
from transformers import AutoModelForSequenceClassification, AutoTokenizer
from torch.optim import AdamW
from sklearn.model_selection import train_test_split
from google.colab import drive
import datetime

# ==========================================
# 1. KẾT NỐI DRIVE & KHỞI TẠO LOGGING
# ==========================================
print("Đang kết nối Google Drive...")
drive.mount('/content/drive')
drive_path = '/content/drive/MyDrive/train-KLCN'

# Tạo thư mục con để lưu log và output
timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
output_dir = os.path.join(drive_path, f"output_phobert_vsmec_{timestamp}")
os.makedirs(output_dir)

# Xóa các handler mặc định của Colab để ghi log không bị trắng
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

log_file = os.path.join(output_dir, 'training_log.txt')
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(message)s',
    handlers=[
        logging.FileHandler(log_file, mode='w', encoding='utf-8'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger()
logger.info(f"Bắt đầu huấn luyện PhoBERT. Output lưu tại: {output_dir}")

# ==========================================
# 2. MODULE TIỀN XỬ LÝ (TỪ FILE CSV TRÊN DRIVE)
# ==========================================
logger.info("Đang nạp từ điển Emoji và Teencode từ Google Drive...")

# Tải từ điển Emoji
try:
    df_emoji = pd.read_csv(os.path.join(drive_path, 'vietnamese_emoji.csv'))
    # Map cột 'emoji' với cột 'description'
    EMOJI_DICT = {str(row['emoji']): str(row['description']).strip() for _, row in df_emoji.iterrows()}
    logger.info(f"✅ Đã nạp thành công {len(EMOJI_DICT)} mô tả Emoji.")
except Exception as e:
    logger.error(f"Lỗi nạp Emoji: {e}")
    EMOJI_DICT = {"😂": "cười", "😍": "yêu thích", "😭": "khóc buồn"}

# Tải từ điển Teencode (Dùng thuật toán trích xuất từ vựng từ câu song ngữ)
try:
    df_teencode = pd.read_csv(os.path.join(drive_path, 'vietnamese_teencode.csv'))
    TEENCODE_DICT = {}

    for _, row in df_teencode.iterrows():
        err_words = str(row['error_text']).lower().split()
        corr_words = str(row['correct_text']).lower().split()

        # Chỉ trích xuất từ lóng nếu 2 câu có cùng số lượng từ (để gióng hàng an toàn)
        if len(err_words) == len(corr_words):
            for e, c in zip(err_words, corr_words):
                if e != c and e not in TEENCODE_DICT:
                    TEENCODE_DICT[e] = c

    logger.info(f"✅ Đã trích xuất thành công {len(TEENCODE_DICT)} từ lóng/Teencode duy nhất.")
except Exception as e:
    logger.error(f"Lỗi nạp Teencode: {e}")
    TEENCODE_DICT = {"khum": "không", "dc": "được", "vs": "với"}

# Hàm Xử Lý Văn Bản Chính
def preprocess_vietnamese_text(text):
    if not isinstance(text, str): return ""

    # Bước A: Dịch Emoji thành văn bản tiếng Việt
    for emo, replacement in EMOJI_DICT.items():
        if emo in text:
            text = text.replace(emo, f" {replacement} ")

    # Xóa ký tự đặc biệt (giữ lại chữ, số và khoảng trắng)
    text = re.sub(r'[^\w\s]', ' ', text)

    # Bước B: Xử lý Teencode bằng thuật toán Tra Từ Điển O(1)
    words = text.split()
    normalized_words = [TEENCODE_DICT.get(w.lower(), w) for w in words]
    text = " ".join(normalized_words)

    # Xóa khoảng trắng thừa
    text = re.sub(r'\s+', ' ', text).strip()

    # Bước C: Tách từ bằng pyvi cho PhoBERT
    text = ViTokenizer.tokenize(text)

    return text.lower()

# ==========================================
# 3. ĐỌC VÀ CHUẨN BỊ DỮ LIỆU UIT-VSMEC
# ==========================================
logger.info("Đang đọc dữ liệu UIT-VSMEC từ Google Drive...")
file_path = os.path.join(drive_path, 'uit_vsmec.csv')
df = pd.read_csv(file_path)[['Sentence', 'Emotion']]

# MAPPER: Chuyển đổi nhãn chữ sang số nguyên cho PyTorch
LABEL_MAP = {
    'Anger': 0,
    'Disgust': 1,
    'Enjoyment': 2,
    'Fear': 3,
    'Other': 4,
    'Sadness': 5,
    'Surprise': 6
}

if df['Emotion'].dtype == object or isinstance(df['Emotion'].iloc[0], str):
    df['Emotion'] = df['Emotion'].map(LABEL_MAP)

# Dọn dẹp dữ liệu rỗng và ép kiểu về số nguyên
df = df.dropna(subset=['Sentence', 'Emotion'])
df['Emotion'] = df['Emotion'].astype(int)

logger.info("Đang chạy module tiền xử lý văn bản (quá trình này mất khoảng 1-2 phút)...")
df['Processed_Sentence'] = df['Sentence'].apply(preprocess_vietnamese_text)

train_texts, val_texts, train_labels, val_labels = train_test_split(
    df['Processed_Sentence'].tolist(),
    df['Emotion'].tolist(),
    test_size=0.1,
    random_state=42
)

# ==========================================
# 4. XÂY DỰNG DATASET & DATALOADER
# ==========================================
class VSMECDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self): return len(self.texts)

    def __getitem__(self, idx):
        text = self.texts[idx]
        label = torch.tensor(self.labels[idx], dtype=torch.long)

        encoding = self.tokenizer(
            text, add_special_tokens=True, max_length=self.max_len,
            padding='max_length', truncation=True, return_tensors='pt'
        )
        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': label
        }

logger.info("Đang tải Tokenizer của PhoBERT...")
tokenizer = AutoTokenizer.from_pretrained("vinai/phobert-base-v2")
train_dataset = VSMECDataset(train_texts, train_labels, tokenizer)
val_dataset = VSMECDataset(val_texts, val_labels, tokenizer)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=16)

# ==========================================
# 5. KHỞI TẠO MÔ HÌNH VÀ TỐI ƯU HÓA
# ==========================================
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
logger.info(f"Khởi tạo mô hình trên thiết bị: {device}")

model = AutoModelForSequenceClassification.from_pretrained("vinai/phobert-base-v2", num_labels=7).to(device)
optimizer = AdamW(model.parameters(), lr=2e-5)
metrics_history = []

# ==========================================
# 6. VÒNG LẶP HUẤN LUYỆN
# ==========================================
epochs = 3
for epoch in range(epochs):
    model.train()
    total_loss = 0
    correct = 0
    total = 0

    for step, batch in enumerate(train_loader):
        optimizer.zero_grad()
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        total_loss += loss.item()

        loss.backward()
        optimizer.step()

        predictions = torch.argmax(outputs.logits, dim=-1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

        if step % 50 == 0 and step > 0:
            logger.info(f"Epoch [{epoch+1}/{epochs}] | Step [{step}/{len(train_loader)}] | Train Loss: {total_loss/step:.4f} | Acc: {correct/total:.4f}")

    # Validation
    model.eval()
    val_loss, val_correct, val_total = 0, 0, 0
    with torch.no_grad():
        for batch in val_loader:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)

            outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
            val_loss += outputs.loss.item()
            predictions = torch.argmax(outputs.logits, dim=-1)
            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

    avg_val_loss = val_loss/len(val_loader)
    avg_val_acc = val_correct/val_total

    logger.info(f"--- TỔNG KẾT EPOCH {epoch+1} ---")
    logger.info(f"Val Loss: {avg_val_loss:.4f} | Val Accuracy: {avg_val_acc:.4f}\n")

    metrics_history.append({
        "epoch": epoch + 1,
        "val_loss": avg_val_loss,
        "val_accuracy": avg_val_acc
    })

# Lưu mô hình
model_save_path = os.path.join(output_dir, "phobert_vsmec_classifier.pt")
torch.save(model.state_dict(), model_save_path)
logger.info(f"Đã lưu mô hình PhoBERT vào: {model_save_path}")

metrics_file = os.path.join(output_dir, 'training_metrics.json')
with open(metrics_file, 'w') as f:
    json.dump(metrics_history, f, indent=4)

# ==========================================
# 7. LOGIC NỘI SUY TỌA ĐỘ V-A CHO BACKEND
# ==========================================
logger.info("=========================================")
logger.info("TEST HÀM NỘI SUY TỌA ĐỘ CHO BACKEND")

def predict_va_coordinate(text):
    model.eval()
    proc_text = preprocess_vietnamese_text(text)
    inputs = tokenizer(proc_text, return_tensors="pt", max_length=128, padding='max_length', truncation=True).to(device)

    with torch.no_grad():
        outputs = model(**inputs)
        probs = torch.softmax(outputs.logits, dim=-1).squeeze().cpu().numpy()

    VA_ANCHORS = {
        0: [-0.8,  0.8], # Anger
        1: [-0.6,  0.4], # Disgust
        2: [ 0.8,  0.7], # Enjoyment
        3: [-0.7,  0.9], # Fear
        4: [ 0.0,  0.0], # Other
        5: [-0.8, -0.6], # Sadness
        6: [ 0.5,  0.9]  # Surprise
    }

    final_valence = sum(probs[i] * VA_ANCHORS[i][0] for i in range(7))
    final_arousal = sum(probs[i] * VA_ANCHORS[i][1] for i in range(7))

    return final_valence, final_arousal

test_sentences = [
    "Phim này cuốn dã man, coi cười đau cả bụng 😂",
    "Chán vcl, hôm nay toàn chuyện xui xẻo chằm zn",
    "Sợ quá đi mất, giật cả mình 😱"
]

for sentence in test_sentences:
    v, a = predict_va_coordinate(sentence)
    logger.info(f"Câu gốc: '{sentence}'")
    logger.info(f" -> Tọa độ V-A: Valence = {v:.4f} | Arousal = {a:.4f}\n")
```

```python
# ============================================================
# PhoBERT - UIT-VSMEC + tiền xử lý Emoji/Teencode theo 3 CSV của bạn
# Chạy trên Google Colab
# ============================================================

# Nếu thiếu thư viện, chạy cell riêng:
# !pip install -q transformers sentencepiece pyvi scikit-learn pandas

import os
import re
import json
import logging
import datetime
import random
from collections import Counter, defaultdict

import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoModelForSequenceClassification, AutoTokenizer
from torch.optim import AdamW
from sklearn.model_selection import train_test_split
from pyvi import ViTokenizer
from google.colab import drive

# ============================================================
# 0. CẤU HÌNH
# ============================================================
SEED = 42
MODEL_NAME = "vinai/phobert-base-v2"
DRIVE_PATH = "/content/drive/MyDrive/train-KLCN"
CSV_NAME = "uit_vsmec.csv"
EMOJI_CSV = "vietnamese_emoji.csv"
TEENCODE_CSV = "vietnamese_teencode.csv"

MAX_LEN = 128
BATCH_SIZE = 16
EPOCHS = 3
LEARNING_RATE = 2e-5
NUM_LABELS = 7

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Nhãn được cố định theo đúng thứ tự VA_ANCHORS ở phần cuối
LABEL2ID = {
    "anger": 0,
    "disgust": 1,
    "enjoyment": 2,
    "fear": 3,
    "other": 4,
    "sadness": 5,
    "surprise": 6,
}
ID2LABEL = {v: k.title() for k, v in LABEL2ID.items()}

# ============================================================
# 1. KẾT NỐI GOOGLE DRIVE VÀ LOGGING
# ============================================================
print("Đang kết nối Google Drive...")
drive.mount("/content/drive")

if not os.path.isdir(DRIVE_PATH):
    raise FileNotFoundError(
        f"Không tìm thấy thư mục dữ liệu: {DRIVE_PATH}\n"
        "Hãy kiểm tra tên thư mục trong Google Drive."
    )

timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
OUTPUT_DIR = os.path.join(DRIVE_PATH, f"output_phobert_vsmec_{timestamp}")
os.makedirs(OUTPUT_DIR, exist_ok=True)

for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    handlers=[
        logging.FileHandler(
            os.path.join(OUTPUT_DIR, "training_log.txt"),
            mode="w",
            encoding="utf-8",
        ),
        logging.StreamHandler(),
    ],
)
logger = logging.getLogger(__name__)
logger.info("Bắt đầu huấn luyện PhoBERT.")
logger.info("Thư mục output: %s", OUTPUT_DIR)

# ============================================================
# 2. NẠP TỪ ĐIỂN EMOJI VÀ TEENCODE
# ============================================================
def load_two_column_dictionary(path, lowercase_keys=False):
    """Đọc từ điển từ hai cột đầu tiên của CSV."""
    if not os.path.isfile(path):
        raise FileNotFoundError(f"Không tìm thấy file từ điển: {path}")

    dictionary_df = pd.read_csv(path, encoding="utf-8-sig")
    if dictionary_df.shape[1] < 2:
        raise ValueError(
            f"File {os.path.basename(path)} cần ít nhất 2 cột, "
            f"nhưng hiện có {dictionary_df.shape[1]} cột."
        )

    dictionary = {}
    for key, value in zip(dictionary_df.iloc[:, 0], dictionary_df.iloc[:, 1]):
        if pd.isna(key) or pd.isna(value):
            continue
        key = str(key).strip()
        value = str(value).strip()
        if not key or not value:
            continue
        if lowercase_keys:
            key = key.lower()
        dictionary[key] = value
    return dictionary


try:
    EMOJI_DICT = load_two_column_dictionary(
        os.path.join(DRIVE_PATH, EMOJI_CSV)
    )
    logger.info("Đã nạp %d mục Emoji.", len(EMOJI_DICT))
except Exception as e:
    logger.warning("Không nạp được từ điển Emoji: %s", e)
    EMOJI_DICT = {
        "😂": "cười",
        "😍": "yêu thích",
        "😭": "khóc buồn",
        "😱": "sợ hãi",
    }

def build_teencode_dictionary(path, min_count=5, min_confidence=0.90):
    """    File vietnamese_teencode.csv là dữ liệu câu song song:
      correct_text = câu đúng
      error_text   = câu có lỗi

    Không thể đọc thẳng cả câu thành từ điển từ-vựng. Hàm này rút trích
    ánh xạ lỗi-từ -> từ-đúng từ các cặp câu có cùng số token, sau đó chỉ
    giữ các ánh xạ đủ số lần xuất hiện và đủ độ tin cậy.
    """
    if not os.path.isfile(path):
        raise FileNotFoundError(f"Không tìm thấy file Teencode: {path}")

    pair_df = pd.read_csv(path, encoding="utf-8-sig")
    required = {"correct_text", "error_text"}
    missing = required - set(pair_df.columns)
    if missing:
        raise ValueError(
            f"File Teencode thiếu cột {sorted(missing)}. "
            f"Các cột hiện có: {pair_df.columns.tolist()}"
        )

    pair_df = pair_df[["correct_text", "error_text"]].dropna().copy()
    pair_df["correct_text"] = pair_df["correct_text"].astype(str).str.lower().str.strip()
    pair_df["error_text"] = pair_df["error_text"].astype(str).str.lower().str.strip()

    # Mỗi error token sẽ bỏ phiếu cho token đúng tương ứng.
    token_votes = defaultdict(Counter)
    token_pattern = re.compile(r"^\w+$", flags=re.UNICODE)
    same_length_pairs = 0

    for correct_text, error_text in zip(
        pair_df["correct_text"], pair_df["error_text"]
    ):
        correct_tokens = correct_text.split()
        error_tokens = error_text.split()

        # Chỉ căn chỉnh vị trí khi hai câu có cùng số token, tránh lệch vị trí.
        if len(correct_tokens) != len(error_tokens):
            continue
        same_length_pairs += 1

        for correct_token, error_token in zip(correct_tokens, error_tokens):
            if correct_token == error_token:
                continue
            if not token_pattern.fullmatch(correct_token):
                continue
            if not token_pattern.fullmatch(error_token):
                continue
            token_votes[error_token][correct_token] += 1

    dictionary = {}
    for error_token, targets in token_votes.items():
        correct_token, count = targets.most_common(1)[0]
        total_votes = sum(targets.values())
        confidence = count / total_votes
        if count >= min_count and confidence >= min_confidence:
            dictionary[error_token] = correct_token

    logger.info(
        "Teencode: %d cặp câu hợp lệ sau khi bỏ dòng thiếu; "
        "%d cặp có cùng số token; tạo được %d ánh xạ từ lỗi -> từ đúng "
        "(min_count=%d, min_confidence=%.0f%%).",
        len(pair_df), same_length_pairs, len(dictionary),
        min_count, min_confidence * 100,
    )
    if not dictionary:
        raise ValueError(
            "Không rút trích được từ điển Teencode. "
            "Hãy kiểm tra định dạng vietnamese_teencode.csv."
        )
    return dictionary


TEENCODE_DICT = build_teencode_dictionary(
    os.path.join(DRIVE_PATH, TEENCODE_CSV),
    min_count=5,
    min_confidence=0.90,
)


def preprocess_vietnamese_text(text):
    """Chuẩn hóa văn bản, thay emoji/teencode và tách từ bằng PyVi."""
    if pd.isna(text) or not isinstance(text, str):
        return ""

    # Thay emoji trước khi loại ký tự đặc biệt
    for emoji, replacement in EMOJI_DICT.items():
        if emoji and emoji in text:
            text = text.replace(emoji, f" {replacement} ")

    # Giữ chữ, số, dấu gạch dưới và khoảng trắng
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Chuẩn hóa teencode theo từng từ
    words = text.split()
    words = [TEENCODE_DICT.get(word.lower(), word) for word in words]
    text = " ".join(words)
    text = re.sub(r"\s+", " ", text).strip()

    if not text:
        return ""

    # PhoBERT cần văn bản tiếng Việt đã tách từ
    text = ViTokenizer.tokenize(text)
    return text.lower()


# ============================================================
# 3. ĐỌC VÀ KIỂM TRA DỮ LIỆU UIT-VSMEC
# ============================================================
logger.info("Đang đọc dữ liệu UIT-VSMEC...")
CSV_PATH = os.path.join(DRIVE_PATH, CSV_NAME)

if not os.path.isfile(CSV_PATH):
    raise FileNotFoundError(
        f"Không tìm thấy file dữ liệu: {CSV_PATH}\n"
        "Hãy đặt uit_vsmec.csv trong thư mục train-KLCN."
    )

# utf-8-sig giúp xử lý CSV có BOM; thử thêm encoding phổ biến nếu cần
try:
    df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")
except UnicodeDecodeError:
    df = pd.read_csv(CSV_PATH, encoding="cp1258")

# Chuẩn hóa tên cột để tránh lỗi do khoảng trắng
df.columns = [str(c).strip() for c in df.columns]

required_columns = {"Sentence", "Emotion"}
missing_columns = required_columns - set(df.columns)
if missing_columns:
    raise ValueError(
        f"CSV thiếu cột {sorted(missing_columns)}. "
        f"Các cột hiện có: {df.columns.tolist()}"
    )

df = df[["Sentence", "Emotion"]].copy()
df["Sentence"] = df["Sentence"].fillna("").astype(str).str.strip()
df["Emotion"] = df["Emotion"].fillna("").astype(str).str.strip().str.lower()

# Bỏ mẫu thiếu câu hoặc nhãn
df = df[
    df["Sentence"].ne("") &
    df["Emotion"].ne("") &
    df["Emotion"].ne("nan")
].copy()

if df.empty:
    raise ValueError("Không còn mẫu hợp lệ sau khi loại dữ liệu thiếu.")

# Không tự ý chuyển nhãn lạ thành Other vì sẽ làm sai nhãn dữ liệu
unknown_labels = sorted(set(df["Emotion"].unique()) - set(LABEL2ID.keys()))
if unknown_labels:
    raise ValueError(
        "Cột Emotion có nhãn chưa được ánh xạ: "
        f"{unknown_labels}\n"
        "Hãy kiểm tra dữ liệu CSV và cập nhật LABEL2ID đúng theo bộ dữ liệu."
    )

df["label_id"] = df["Emotion"].map(LABEL2ID).astype("int64")

# Bỏ các câu rỗng sau tiền xử lý
logger.info("Đang tiền xử lý văn bản...")
df["Processed_Sentence"] = df["Sentence"].apply(preprocess_vietnamese_text)
df = df[df["Processed_Sentence"].str.strip().ne("")].copy()

if df.empty:
    raise ValueError("Tất cả văn bản đều rỗng sau tiền xử lý.")

class_counts = df["label_id"].value_counts()
if len(class_counts) != NUM_LABELS:
    logger.warning(
        "Dữ liệu hiện có %d/7 lớp. Các lớp hiện có: %s",
        len(class_counts),
        {ID2LABEL[int(k)]: int(v) for k, v in class_counts.items()},
    )

# Stratify yêu cầu mỗi lớp có tối thiểu 2 mẫu và tập validation đủ chỗ cho các lớp
num_classes_present = df["label_id"].nunique()
validation_size = int(np.ceil(len(df) * 0.1))
can_stratify = (
    class_counts.min() >= 2
    and validation_size >= num_classes_present
    and (len(df) - validation_size) >= num_classes_present
)

if can_stratify:
    stratify_values = df["label_id"]
    logger.info("Chia train/validation có stratify theo nhãn.")
else:
    stratify_values = None
    logger.warning(
        "Không thể stratify an toàn do số mẫu/lớp hoặc kích thước dữ liệu; "
        "chia ngẫu nhiên với random_state=%d.",
        SEED,
    )

train_texts, val_texts, train_labels, val_labels = train_test_split(
    df["Processed_Sentence"].tolist(),
    df["label_id"].astype(int).tolist(),
    test_size=0.1,
    random_state=SEED,
    stratify=stratify_values,
)

logger.info("Tổng số mẫu: %d", len(df))
logger.info("Số mẫu train: %d", len(train_texts))
logger.info("Số mẫu validation: %d", len(val_texts))
logger.info(
    "Phân bố nhãn: %s",
    {ID2LABEL[int(k)]: int(v) for k, v in df["label_id"].value_counts().items()},
)
logger.info("LABEL2ID: %s", LABEL2ID)

# Lưu ánh xạ nhãn để backend sử dụng thống nhất
with open(os.path.join(OUTPUT_DIR, "label_mapping.json"), "w", encoding="utf-8") as f:
    json.dump(
        {"label2id": LABEL2ID, "id2label": ID2LABEL},
        f,
        ensure_ascii=False,
        indent=2,
    )

# ============================================================
# 4. TOKENIZER, DATASET VÀ DATALOADER
# ============================================================
logger.info("Đang tải tokenizer PhoBERT...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)


class VSMECDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        if len(texts) != len(labels):
            raise ValueError(
                f"Số văn bản ({len(texts)}) khác số nhãn ({len(labels)})."
            )

        self.texts = list(texts)
        self.labels = list(labels)
        self.tokenizer = tokenizer
        self.max_len = max_len

        for i, label in enumerate(self.labels):
            if isinstance(label, (bool, np.bool_)) or not isinstance(
                label, (int, np.integer)
            ):
                raise TypeError(
                    f"Nhãn tại vị trí {i} phải là số nguyên, nhận được "
                    f"{label!r} ({type(label).__name__}). "
                    "Hãy kiểm tra bước LABEL2ID."
                )
            if not 0 <= int(label) < NUM_LABELS:
                raise ValueError(f"Nhãn ngoài khoảng 0-{NUM_LABELS - 1}: {label}")

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts[idx])
        label = torch.tensor(int(self.labels[idx]), dtype=torch.long)

        encoding = self.tokenizer(
            text,
            add_special_tokens=True,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_tensors="pt",
        )

        return {
            "input_ids": encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "labels": label,
        }


train_dataset = VSMECDataset(train_texts, train_labels, tokenizer, MAX_LEN)
val_dataset = VSMECDataset(val_texts, val_labels, tokenizer, MAX_LEN)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
)
val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
)

# Kiểm tra một batch trước khi tải mô hình
sample_batch = next(iter(train_loader))
assert sample_batch["input_ids"].dtype == torch.long
assert sample_batch["labels"].dtype == torch.long
assert sample_batch["labels"].min().item() >= 0
assert sample_batch["labels"].max().item() < NUM_LABELS
logger.info("Kiểm tra Dataset/DataLoader thành công.")

# ============================================================
# 5. KHỞI TẠO MÔ HÌNH VÀ OPTIMIZER
# ============================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
logger.info("Thiết bị huấn luyện: %s", device)
if device.type == "cuda":
    logger.info("GPU: %s", torch.cuda.get_device_name(0))
else:
    logger.warning(
        "Đang chạy bằng CPU; PhoBERT có thể huấn luyện rất chậm. "
        "Trong Colab, chọn Runtime > Change runtime type > GPU."
    )

logger.info("Đang tải mô hình %s...", MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=NUM_LABELS,
    id2label=ID2LABEL,
    label2id={v: k for k, v in ID2LABEL.items()},
)
model.to(device)

optimizer = AdamW(model.parameters(), lr=LEARNING_RATE)
metrics_history = []

# ============================================================
# 6. HUẤN LUYỆN VÀ ĐÁNH GIÁ
# ============================================================
if len(train_loader) == 0:
    raise ValueError("Train DataLoader rỗng.")
if len(val_loader) == 0:
    raise ValueError("Validation DataLoader rỗng.")

for epoch in range(EPOCHS):
    model.train()
    running_loss = 0.0
    train_correct = 0
    train_total = 0

    logger.info("========== EPOCH %d/%d ==========", epoch + 1, EPOCHS)

    for step, batch in enumerate(train_loader, start=1):
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["labels"].to(device)

        optimizer.zero_grad(set_to_none=True)

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels,
        )
        loss = outputs.loss

        if not torch.isfinite(loss):
            raise RuntimeError(
                f"Loss không hữu hạn tại epoch {epoch + 1}, step {step}: {loss.item()}"
            )

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        batch_size_actual = labels.size(0)
        running_loss += loss.item() * batch_size_actual
        predictions = torch.argmax(outputs.logits.detach(), dim=-1)
        train_correct += (predictions == labels).sum().item()
        train_total += batch_size_actual

        if step % 50 == 0 or step == len(train_loader):
            logger.info(
                "Epoch [%d/%d] Step [%d/%d] | Train Loss: %.4f | Train Acc: %.4f",
                epoch + 1,
                EPOCHS,
                step,
                len(train_loader),
                running_loss / max(train_total, 1),
                train_correct / max(train_total, 1),
            )

    train_loss = running_loss / max(train_total, 1)
    train_accuracy = train_correct / max(train_total, 1)

    # Validation
    model.eval()
    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for batch in val_loader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].to(device)

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels,
            )

            batch_size_actual = labels.size(0)
            val_loss_sum += outputs.loss.item() * batch_size_actual
            predictions = torch.argmax(outputs.logits, dim=-1)
            val_correct += (predictions == labels).sum().item()
            val_total += batch_size_actual

    avg_val_loss = val_loss_sum / max(val_total, 1)
    avg_val_acc = val_correct / max(val_total, 1)

    epoch_metrics = {
        "epoch": epoch + 1,
        "train_loss": float(train_loss),
        "train_accuracy": float(train_accuracy),
        "val_loss": float(avg_val_loss),
        "val_accuracy": float(avg_val_acc),
    }
    metrics_history.append(epoch_metrics)

    logger.info("--- TỔNG KẾT EPOCH %d ---", epoch + 1)
    logger.info(
        "Train Loss: %.4f | Train Accuracy: %.4f | Val Loss: %.4f | Val Accuracy: %.4f",
        train_loss,
        train_accuracy,
        avg_val_loss,
        avg_val_acc,
    )

    # Lưu metrics sau mỗi epoch để tránh mất kết quả nếu runtime bị ngắt
    with open(
        os.path.join(OUTPUT_DIR, "training_metrics.json"),
        "w",
        encoding="utf-8",
    ) as f:
        json.dump(metrics_history, f, ensure_ascii=False, indent=2)

# ============================================================
# 7. LƯU MÔ HÌNH VÀ TOKENIZER
# ============================================================
FINAL_MODEL_DIR = os.path.join(OUTPUT_DIR, "phobert_vsmec_model")
os.makedirs(FINAL_MODEL_DIR, exist_ok=True)

model.save_pretrained(FINAL_MODEL_DIR)
tokenizer.save_pretrained(FINAL_MODEL_DIR)

# Lưu thêm state_dict để tương thích với cách nạp trọng số thủ công
state_dict_path = os.path.join(OUTPUT_DIR, "phobert_vsmec_classifier.pt")
torch.save(model.state_dict(), state_dict_path)

logger.info("Đã lưu model/tokenizer tại: %s", FINAL_MODEL_DIR)
logger.info("Đã lưu state_dict tại: %s", state_dict_path)

# ============================================================
# 8. NỘI SUY TỌA ĐỘ VALENCE-AROUSAL (V-A)
# ============================================================
# Thứ tự ID phải khớp LABEL2ID:
# 0 Anger, 1 Disgust, 2 Enjoyment, 3 Fear,
# 4 Other, 5 Sadness, 6 Surprise
VA_ANCHORS = {
    0: [-0.8,  0.8],  # Anger
    1: [-0.6,  0.4],  # Disgust
    2: [ 0.8,  0.7],  # Enjoyment
    3: [-0.7,  0.9],  # Fear
    4: [ 0.0,  0.0],  # Other
    5: [-0.8, -0.6],  # Sadness
    6: [ 0.5,  0.9],  # Surprise
}


def predict_va_coordinate(text):
    """Trả về (valence, arousal, nhãn dự đoán, xác suất)."""
    if not isinstance(text, str) or not text.strip():
        raise ValueError("Văn bản dự đoán không được để trống.")

    model.eval()
    processed_text = preprocess_vietnamese_text(text)

    inputs = tokenizer(
        processed_text,
        return_tensors="pt",
        max_length=MAX_LEN,
        padding="max_length",
        truncation=True,
    )
    inputs = {key: value.to(device) for key, value in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)
        probs = torch.softmax(outputs.logits, dim=-1)[0].cpu().numpy()

    predicted_id = int(np.argmax(probs))
    final_valence = float(
        sum(float(probs[i]) * VA_ANCHORS[i][0] for i in range(NUM_LABELS))
    )
    final_arousal = float(
        sum(float(probs[i]) * VA_ANCHORS[i][1] for i in range(NUM_LABELS))
    )

    return final_valence, final_arousal, ID2LABEL[predicted_id], probs.tolist()


# ============================================================
# 9. KIỂM TRA THỬ
# ============================================================
logger.info("=========================================")
logger.info("TEST THỬ HÀM DỰ ĐOÁN TỌA ĐỘ V-A")

test_sentences = [
    "Phim này cuốn dã man, coi cười đau cả bụng 😂",
    "Chán vcl, hôm nay toàn chuyện xui xẻo chằm zn",
    "Sợ quá đi mất, giật cả mình 😱",
]

for sentence in test_sentences:
    valence, arousal, predicted_label, probabilities = predict_va_coordinate(sentence)
    logger.info("Câu gốc: %s", sentence)
    logger.info("Nhãn dự đoán: %s", predicted_label)
    logger.info(
        "Tọa độ V-A: Valence = %.4f | Arousal = %.4f",
        valence,
        arousal,
    )

logger.info("Hoàn tất toàn bộ quy trình.")
logger.info("Các file kết quả nằm tại: %s", OUTPUT_DIR)
```

```python
!pip install -q transformers sentencepiece pyvi scikit-learn pandas
# ============================================================
# PHOBERT - PHÂN LOẠI 7 CẢM XÚC UIT-VSMEC + TỌA ĐỘ V-A
# Dành cho Google Colab và 3 file CSV của bạn.
# Mục tiêu: cải thiện Validation Accuracy, chọn checkpoint tốt nhất,
# đánh giá độc lập trên test và khảo sát ngưỡng dự đoán >= 80% accuracy.
# LƯU Ý: Không thể bảo đảm mô hình đạt 80% accuracy tổng thể.
# ============================================================

# Nếu thiếu thư viện, chạy trong một cell Colab riêng:
# !pip install -q transformers sentencepiece pyvi scikit-learn pandas

import os
import re
import json
import math
import logging
import datetime
import random
from collections import Counter, defaultdict

import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer,
    get_linear_schedule_with_warmup,
)
from torch.optim import AdamW
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
)
from pyvi import ViTokenizer
from google.colab import drive
# ============================================================
# 0. CẤU HÌNH
# ============================================================
SEED = 42
MODEL_NAME = "vinai/phobert-base-v2"
DRIVE_PATH = "/content/drive/MyDrive/train-KLCN"
CSV_NAME = "uit_vsmec.csv"
EMOJI_CSV = "vietnamese_emoji.csv"
TEENCODE_CSV = "vietnamese_teencode.csv"

MAX_LEN = 128
BATCH_SIZE = 16
EPOCHS = 8
LEARNING_RATE = 1e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.10
EARLY_STOPPING_PATIENCE = 2
NUM_LABELS = 7

# Ngưỡng mục tiêu dùng để khảo sát dự đoán có chọn lọc, KHÔNG phải
# lời hứa accuracy tổng thể. Threshold được tìm trên validation rồi
# kiểm tra lại trên test độc lập.
TARGET_SELECTIVE_ACCURACY = 0.80
MIN_SELECTIVE_COVERAGE = 0.10
MIN_SELECTIVE_SAMPLES = 30

# Nếu muốn so sánh ảnh hưởng của từ điển Teencode, đổi thành False.
USE_TEENCODE = True
USE_EMOJI_DICTIONARY = True

LABEL2ID = {
    "anger": 0,
    "disgust": 1,
    "enjoyment": 2,
    "fear": 3,
    "other": 4,
    "sadness": 5,
    "surprise": 6,
}
ID2LABEL = {
    0: "Anger",
    1: "Disgust",
    2: "Enjoyment",
    3: "Fear",
    4: "Other",
    5: "Sadness",
    6: "Surprise",
}

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
# ============================================================
# 1. GOOGLE DRIVE, THƯ MỤC OUTPUT VÀ LOGGING
# ============================================================
print("Đang kết nối Google Drive...")
drive.mount("/content/drive")

if not os.path.isdir(DRIVE_PATH):
    raise FileNotFoundError(
        f"Không tìm thấy thư mục: {DRIVE_PATH}. "
        "Hãy đặt 3 file CSV trong MyDrive/train-KLCN."
    )

missing_files = [
    name for name in (CSV_NAME, EMOJI_CSV, TEENCODE_CSV)
    if not os.path.isfile(os.path.join(DRIVE_PATH, name))
]
if missing_files:
    raise FileNotFoundError(
        f"Thiếu file trong {DRIVE_PATH}: {missing_files}"
    )

timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
OUTPUT_DIR = os.path.join(DRIVE_PATH, f"output_phobert_target80_{timestamp}")
os.makedirs(OUTPUT_DIR, exist_ok=True)

for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    handlers=[
        logging.FileHandler(
            os.path.join(OUTPUT_DIR, "training_log.txt"),
            mode="w",
            encoding="utf-8",
        ),
        logging.StreamHandler(),
    ],
)
logger = logging.getLogger("phobert_vsmec")
logger.info("Bắt đầu huấn luyện PhoBERT UIT-VSMEC")
logger.info("Output: %s", OUTPUT_DIR)
logger.info("Device sẽ được xác định sau khi đọc dữ liệu.")
# ============================================================
# 2. ĐỌC TỪ ĐIỂN EMOJI VÀ TEENCODE
# ============================================================
def load_emoji_dictionary(path):
    """File emoji có cột emoji và description."""
    df_emoji = pd.read_csv(path, encoding="utf-8-sig")
    required = {"emoji", "description"}
    if not required.issubset(set(df_emoji.columns)):
        raise ValueError(
            f"File emoji cần cột {required}; hiện có {df_emoji.columns.tolist()}"
        )
    result = {}
    for emoji, description in zip(df_emoji["emoji"], df_emoji["description"]):
        if pd.isna(emoji) or pd.isna(description):
            continue
        emoji = str(emoji).strip()
        description = str(description).strip()
        if emoji and description:
            result[emoji] = description
    # Thay chuỗi emoji dài trước để tránh khớp một phần trước.
    return dict(sorted(result.items(), key=lambda item: len(item[0]), reverse=True))


def build_teencode_dictionary(path, min_count=5, min_confidence=0.90):
    """Tạo map error token -> correct token từ cặp câu trong CSV.

    File gốc có cột correct_text và error_text. Chỉ dùng các cặp câu
    cùng số token để căn chỉnh theo vị trí; chỉ giữ ánh xạ có đủ phiếu
    và độ đồng thuận cao nhằm hạn chế sửa sai từ.
    """
    pair_df = pd.read_csv(path, encoding="utf-8-sig")
    required = {"correct_text", "error_text"}
    if not required.issubset(set(pair_df.columns)):
        raise ValueError(
            f"File teencode cần cột {required}; hiện có {pair_df.columns.tolist()}"
        )

    pair_df = pair_df[["correct_text", "error_text"]].dropna().copy()
    pair_df["correct_text"] = pair_df["correct_text"].astype(str).str.lower().str.strip()
    pair_df["error_text"] = pair_df["error_text"].astype(str).str.lower().str.strip()

    votes = defaultdict(Counter)
    token_pattern = re.compile(r"^\w+$", flags=re.UNICODE)
    aligned_pairs = 0

    # CSV có hơn 200 nghìn dòng; đọc tuần tự trên bộ nhớ là vừa phải.
    for correct_text, error_text in zip(pair_df["correct_text"], pair_df["error_text"]):
        correct_tokens = correct_text.split()
        error_tokens = error_text.split()
        if not correct_tokens or len(correct_tokens) != len(error_tokens):
            continue
        aligned_pairs += 1
        for correct_token, error_token in zip(correct_tokens, error_tokens):
            if correct_token == error_token:
                continue
            if not token_pattern.fullmatch(correct_token):
                continue
            if not token_pattern.fullmatch(error_token):
                continue
            votes[error_token][correct_token] += 1

    result = {}
    for error_token, choices in votes.items():
        correct_token, count = choices.most_common(1)[0]
        total = sum(choices.values())
        confidence = count / total
        if count >= min_count and confidence >= min_confidence:
            result[error_token] = correct_token

    logger.info(
        "Teencode: %d dòng hợp lệ, %d cặp cùng số token, %d ánh xạ lỗi->đúng "
        "(min_count=%d, confidence>=%.0f%%).",
        len(pair_df), aligned_pairs, len(result), min_count, min_confidence * 100,
    )
    if not result:
        logger.warning("Không trích xuất được map teencode; bỏ qua bước này.")
    return result


try:
    EMOJI_DICT = load_emoji_dictionary(os.path.join(DRIVE_PATH, EMOJI_CSV))
    logger.info("Đã nạp %d ánh xạ emoji.", len(EMOJI_DICT))
except Exception as exc:
    logger.warning("Không nạp được emoji dictionary: %s", exc)
    EMOJI_DICT = {"😂": "cười", "😍": "yêu thích", "😭": "khóc buồn", "😱": "sợ hãi"}

if USE_TEENCODE:
    TEENCODE_DICT = build_teencode_dictionary(
        os.path.join(DRIVE_PATH, TEENCODE_CSV),
        min_count=5,
        min_confidence=0.90,
    )
else:
    TEENCODE_DICT = {}
    logger.info("USE_TEENCODE=False: bỏ qua chuẩn hóa teencode.")


def preprocess_vietnamese_text(text):
    """Chuẩn hóa emoji, teencode, khoảng trắng và tách từ bằng PyVi."""
    if not isinstance(text, str) or not text.strip():
        return ""

    if USE_EMOJI_DICTIONARY and EMOJI_DICT:
        for emoji, description in EMOJI_DICT.items():
            if emoji in text:
                text = text.replace(emoji, f" {description} ")

    # Giữ chữ/số/dấu gạch dưới và khoảng trắng.
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)
    words = text.lower().split()
    if TEENCODE_DICT:
        words = [TEENCODE_DICT.get(word, word) for word in words]
    text = re.sub(r"\s+", " ", " ".join(words)).strip()

    if not text:
        return ""
    return ViTokenizer.tokenize(text).lower()
    # ============================================================
# 3. ĐỌC VÀ TIỀN XỬ LÝ UIT-VSMEC
# ============================================================
CSV_PATH = os.path.join(DRIVE_PATH, CSV_NAME)
logger.info("Đọc dữ liệu: %s", CSV_PATH)
df = pd.read_csv(CSV_PATH, encoding="utf-8-sig")
df.columns = [str(col).strip() for col in df.columns]
required_columns = {"Sentence", "Emotion"}
missing = required_columns - set(df.columns)
if missing:
    raise ValueError(
        f"uit_vsmec.csv thiếu cột {sorted(missing)}; có {df.columns.tolist()}"
    )

df = df[["Sentence", "Emotion"]].copy()
df["Sentence"] = df["Sentence"].fillna("").astype(str).str.strip()
df["Emotion"] = df["Emotion"].fillna("").astype(str).str.strip().str.lower()
df = df[
    df["Sentence"].ne("") & df["Emotion"].ne("") & df["Emotion"].ne("nan")
].copy()
if df.empty:
    raise ValueError("Không có mẫu dữ liệu hợp lệ trong UIT-VSMEC.")

unknown_labels = sorted(set(df["Emotion"].unique()) - set(LABEL2ID))
if unknown_labels:
    raise ValueError(
        f"Nhãn chưa được ánh xạ: {unknown_labels}. Kiểm tra cột Emotion."
    )
df["label_id"] = df["Emotion"].map(LABEL2ID).astype("int64")

logger.info("Tiền xử lý %d câu văn...", len(df))
df["Processed_Sentence"] = df["Sentence"].apply(preprocess_vietnamese_text)
df = df[df["Processed_Sentence"].str.strip().ne("")].copy()
if df.empty:
    raise ValueError("Tất cả câu đều rỗng sau bước tiền xử lý.")

# Chia 80% train / 10% validation / 10% test, stratified theo nhãn.
all_texts = df["Processed_Sentence"].tolist()
all_labels = df["label_id"].astype(int).tolist()

train_texts, temp_texts, train_labels, temp_labels = train_test_split(
    all_texts,
    all_labels,
    test_size=0.20,
    random_state=SEED,
    stratify=all_labels,
)
val_texts, test_texts, val_labels, test_labels = train_test_split(
    temp_texts,
    temp_labels,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_labels,
)

logger.info(
    "Kích thước dữ liệu: train=%d, validation=%d, test=%d",
    len(train_texts), len(val_texts), len(test_texts),
)
logger.info(
    "Phân bố nhãn toàn bộ: %s",
    {ID2LABEL[int(k)]: int(v) for k, v in df["label_id"].value_counts().items()},
)

with open(os.path.join(OUTPUT_DIR, "label_mapping.json"), "w", encoding="utf-8") as f:
    json.dump({"label2id": LABEL2ID, "id2label": ID2LABEL}, f, ensure_ascii=False, indent=2)
# ============================================================
# 4. TOKENIZER, DATASET VÀ DATALOADER
# ============================================================
logger.info("Tải tokenizer %s...", MODEL_NAME)
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)


class VSMECDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        if len(texts) != len(labels):
            raise ValueError("Số văn bản khác số nhãn.")
        self.texts = list(texts)
        self.labels = [int(label) for label in labels]
        self.tokenizer = tokenizer
        self.max_len = max_len
        for i, label in enumerate(self.labels):
            if label < 0 or label >= NUM_LABELS:
                raise ValueError(f"Nhãn tại vị trí {i} ngoài khoảng 0-6: {label}")

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts[idx])
        encoded = self.tokenizer(
            text,
            add_special_tokens=True,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_tensors="pt",
        )
        return {
            "input_ids": encoded["input_ids"].squeeze(0),
            "attention_mask": encoded["attention_mask"].squeeze(0),
            "labels": torch.tensor(self.labels[idx], dtype=torch.long),
        }


train_dataset = VSMECDataset(train_texts, train_labels, tokenizer, MAX_LEN)
val_dataset = VSMECDataset(val_texts, val_labels, tokenizer, MAX_LEN)
test_dataset = VSMECDataset(test_texts, test_labels, tokenizer, MAX_LEN)
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

sample_batch = next(iter(train_loader))
assert sample_batch["input_ids"].dtype == torch.long
assert sample_batch["labels"].dtype == torch.long
assert int(sample_batch["labels"].min()) >= 0 and int(sample_batch["labels"].max()) < NUM_LABELS
logger.info("Dataset/DataLoader kiểm tra thành công.")
# ============================================================
# 5. MODEL, OPTIMIZER VÀ SCHEDULER
# ============================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
logger.info("Thiết bị huấn luyện: %s", device)
if device.type == "cuda":
    logger.info("GPU: %s", torch.cuda.get_device_name(0))
else:
    logger.warning("Đang chạy CPU; PhoBERT có thể rất chậm. Hãy bật GPU trong Colab.")

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=NUM_LABELS,
    id2label=ID2LABEL,
    label2id={label: idx for idx, label in ID2LABEL.items()},
).to(device)

optimizer = AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
total_training_steps = len(train_loader) * EPOCHS
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=max(1, int(total_training_steps * WARMUP_RATIO)),
    num_training_steps=total_training_steps,
)

BEST_MODEL_DIR = os.path.join(OUTPUT_DIR, "best_model")
os.makedirs(BEST_MODEL_DIR, exist_ok=True)
best_val_accuracy = -1.0
best_val_macro_f1 = -1.0
best_epoch = 0
patience_counter = 0
metrics_history = []
# ============================================================
# 6. HÀM ĐÁNH GIÁ
# ============================================================
def evaluate_model(model, loader, collect_probabilities=False):
    model.eval()
    all_labels = []
    all_preds = []
    all_confidences = []
    all_probabilities = []
    loss_sum = 0.0
    total = 0

    with torch.no_grad():
        for batch in loader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["labels"].to(device)
            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                labels=labels,
            )
            batch_size = labels.size(0)
            loss_sum += outputs.loss.item() * batch_size
            probs = torch.softmax(outputs.logits, dim=-1)
            confidences, predictions = probs.max(dim=-1)

            total += batch_size
            all_labels.extend(labels.detach().cpu().tolist())
            all_preds.extend(predictions.detach().cpu().tolist())
            all_confidences.extend(confidences.detach().cpu().tolist())
            if collect_probabilities:
                all_probabilities.extend(probs.detach().cpu().tolist())

    result = {
        "loss": loss_sum / max(total, 1),
        "accuracy": accuracy_score(all_labels, all_preds),
        "macro_f1": f1_score(all_labels, all_preds, average="macro", zero_division=0),
        "labels": all_labels,
        "predictions": all_preds,
        "confidences": all_confidences,
        "probabilities": all_probabilities if collect_probabilities else None,
    }
    return result
    # ============================================================
# 7. HUẤN LUYỆN, LƯU CHECKPOINT TỐT NHẤT, EARLY STOPPING
# ============================================================
for epoch in range(EPOCHS):
    model.train()
    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    logger.info("========== EPOCH %d/%d ==========", epoch + 1, EPOCHS)

    for step, batch in enumerate(train_loader, start=1):
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels = batch["labels"].to(device)

        optimizer.zero_grad(set_to_none=True)
        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=labels,
        )
        loss = outputs.loss
        if not torch.isfinite(loss):
            raise RuntimeError(
                f"Loss không hữu hạn tại epoch={epoch + 1}, step={step}: {loss.item()}"
            )

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        scheduler.step()

        batch_size = labels.size(0)
        train_loss_sum += loss.item() * batch_size
        predictions = torch.argmax(outputs.logits.detach(), dim=-1)
        train_correct += (predictions == labels).sum().item()
        train_total += batch_size

        if step % 50 == 0 or step == len(train_loader):
            logger.info(
                "Epoch [%d/%d] Step [%d/%d] | Train Loss: %.4f | Train Acc: %.4f | LR: %.2e",
                epoch + 1, EPOCHS, step, len(train_loader),
                train_loss_sum / max(train_total, 1),
                train_correct / max(train_total, 1),
                scheduler.get_last_lr()[0],
            )

    train_loss = train_loss_sum / max(train_total, 1)
    train_accuracy = train_correct / max(train_total, 1)
    val_result = evaluate_model(model, val_loader)

    current = {
        "epoch": epoch + 1,
        "train_loss": float(train_loss),
        "train_accuracy": float(train_accuracy),
        "val_loss": float(val_result["loss"]),
        "val_accuracy": float(val_result["accuracy"]),
        "val_macro_f1": float(val_result["macro_f1"]),
    }
    metrics_history.append(current)

    logger.info("--- TỔNG KẾT EPOCH %d ---", epoch + 1)
    logger.info(
        "Train Loss: %.4f | Train Acc: %.4f | Val Loss: %.4f | Val Acc: %.4f | Val Macro F1: %.4f",
        train_loss, train_accuracy, val_result["loss"],
        val_result["accuracy"], val_result["macro_f1"],
    )

    # Chọn checkpoint bằng validation accuracy; khi hòa, ưu tiên Macro F1.
    improved = (
        val_result["accuracy"] > best_val_accuracy
        or (
            math.isclose(val_result["accuracy"], best_val_accuracy, abs_tol=1e-12)
            and val_result["macro_f1"] > best_val_macro_f1
        )
    )

    if improved:
        best_val_accuracy = float(val_result["accuracy"])
        best_val_macro_f1 = float(val_result["macro_f1"])
        best_epoch = epoch + 1
        patience_counter = 0
        model.save_pretrained(BEST_MODEL_DIR)
        tokenizer.save_pretrained(BEST_MODEL_DIR)
        logger.info(
            "Đã lưu checkpoint tốt nhất (epoch %d, val_acc=%.4f, macro_f1=%.4f).",
            best_epoch, best_val_accuracy, best_val_macro_f1,
        )
    else:
        patience_counter += 1
        logger.info(
            "Validation không cải thiện: patience %d/%d",
            patience_counter, EARLY_STOPPING_PATIENCE,
        )

    with open(os.path.join(OUTPUT_DIR, "training_metrics.json"), "w", encoding="utf-8") as f:
        json.dump(metrics_history, f, ensure_ascii=False, indent=2)

    if patience_counter >= EARLY_STOPPING_PATIENCE:
        logger.info("Early stopping sau epoch %d.", epoch + 1)
        break

logger.info("Epoch tốt nhất: %d | Validation Accuracy tốt nhất: %.4f", best_epoch, best_val_accuracy)
# ============================================================
# 8. NẠP CHECKPOINT TỐT NHẤT, ĐÁNH GIÁ TEST ĐỘC LẬP
# ============================================================
logger.info("Nạp checkpoint tốt nhất...")
del model
if torch.cuda.is_available():
    torch.cuda.empty_cache()
model = AutoModelForSequenceClassification.from_pretrained(BEST_MODEL_DIR).to(device)
tokenizer = AutoTokenizer.from_pretrained(BEST_MODEL_DIR)
model.eval()

val_result = evaluate_model(model, val_loader, collect_probabilities=True)
test_result = evaluate_model(model, test_loader, collect_probabilities=True)

logger.info("========== KẾT QUẢ TEST ĐỘC LẬP ==========")
logger.info("Test Loss: %.4f", test_result["loss"])
logger.info("Test Accuracy: %.2f%%", test_result["accuracy"] * 100)
logger.info("Test Macro F1: %.4f", test_result["macro_f1"])

class_names = [ID2LABEL[i] for i in range(NUM_LABELS)]
report = classification_report(
    test_result["labels"],
    test_result["predictions"],
    labels=list(range(NUM_LABELS)),
    target_names=class_names,
    digits=4,
    zero_division=0,
)
print("\n===== CLASSIFICATION REPORT - TEST =====")
print(report)
logger.info("\nClassification Report - Test:\n%s", report)

cm = confusion_matrix(
    test_result["labels"],
    test_result["predictions"],
    labels=list(range(NUM_LABELS)),
)
cm_df = pd.DataFrame(cm, index=class_names, columns=class_names)
cm_path = os.path.join(OUTPUT_DIR, "test_confusion_matrix.csv")
cm_df.to_csv(cm_path, encoding="utf-8-sig")
print("\n===== CONFUSION MATRIX - TEST =====")
print(cm_df)
# ============================================================
# 9. TÌM NGƯỠNG >=80% TRÊN VALIDATION, KIỂM TRA TRÊN TEST
# ============================================================
# Chọn threshold dựa trên validation. Không dùng test để chọn threshold.
def choose_selective_threshold(
    confidences,
    predictions,
    labels,
    target_accuracy=0.80,
    min_coverage=0.10,
    min_samples=30,
):
    conf = np.asarray(confidences, dtype=float)
    preds = np.asarray(predictions, dtype=int)
    true = np.asarray(labels, dtype=int)
    if len(conf) == 0:
        return None

    # Mỗi ngưỡng là một confidence thực tế, nên không bỏ sót điểm chuyển tập con.
    candidates = np.unique(conf)
    candidates = np.sort(candidates)
    valid = []
    for threshold in candidates:
        mask = conf >= threshold
        count = int(mask.sum())
        coverage = count / len(conf)
        if count < min_samples or coverage < min_coverage:
            continue
        accuracy = float((preds[mask] == true[mask]).mean())
        if accuracy >= target_accuracy:
            valid.append({
                "threshold": float(threshold),
                "accuracy": accuracy,
                "coverage": float(coverage),
                "count": count,
            })

    if not valid:
        return None
    # Tối đa coverage trong số các ngưỡng đạt target trên validation.
    return max(valid, key=lambda row: (row["coverage"], row["accuracy"]))

threshold_result = choose_selective_threshold(
    val_result["confidences"],
    val_result["predictions"],
    val_result["labels"],
    target_accuracy=TARGET_SELECTIVE_ACCURACY,
    min_coverage=MIN_SELECTIVE_COVERAGE,
    min_samples=MIN_SELECTIVE_SAMPLES,
)

confidence_threshold = None
if threshold_result is None:
    logger.warning(
        "Không tìm thấy ngưỡng đạt %.0f%% accuracy trên validation với coverage >= %.0f%% "
        "và tối thiểu %d mẫu. Không tạo ngưỡng giả.",
        TARGET_SELECTIVE_ACCURACY * 100,
        MIN_SELECTIVE_COVERAGE * 100,
        MIN_SELECTIVE_SAMPLES,
    )
    print(
        "\nKhông tìm được threshold đáp ứng mục tiêu trên validation. "
        "Không có cơ sở để khẳng định mô hình đạt 80%% accuracy ngay cả trên tập dự đoán được chọn."
    )
else:
    confidence_threshold = threshold_result["threshold"]
    logger.info(
        "Threshold từ validation: %.4f | Val selective accuracy: %.2f%% | Coverage: %.2f%% (%d/%d)",
        confidence_threshold,
        threshold_result["accuracy"] * 100,
        threshold_result["coverage"] * 100,
        threshold_result["count"],
        len(val_result["labels"]),
    )

    test_conf = np.asarray(test_result["confidences"], dtype=float)
    test_preds = np.asarray(test_result["predictions"], dtype=int)
    test_true = np.asarray(test_result["labels"], dtype=int)
    test_mask = test_conf >= confidence_threshold
    test_count = int(test_mask.sum())
    test_coverage = test_count / len(test_conf)
    if test_count:
        test_selective_accuracy = float((test_preds[test_mask] == test_true[test_mask]).mean())
        logger.info(
            "TEST selective: accuracy=%.2f%% | coverage=%.2f%% (%d/%d) | threshold=%.4f",
            test_selective_accuracy * 100,
            test_coverage * 100,
            test_count,
            len(test_conf),
            confidence_threshold,
        )
        print(
            "\n===== SELECTIVE PREDICTION - TEST =====\n"
            f"Threshold đã chọn từ validation: {confidence_threshold:.4f}\n"
            f"Test accuracy trên mẫu được nhận: {test_selective_accuracy:.2%}\n"
            f"Test coverage: {test_coverage:.2%} ({test_count}/{len(test_conf)} mẫu)"
        )
    else:
        test_selective_accuracy = None
        logger.warning("Không có mẫu test nào đạt threshold đã chọn.")

threshold_path = os.path.join(OUTPUT_DIR, "confidence_threshold.json")
with open(threshold_path, "w", encoding="utf-8") as f:
    json.dump({
        "target_selective_accuracy": TARGET_SELECTIVE_ACCURACY,
        "min_selective_coverage": MIN_SELECTIVE_COVERAGE,
        "validation_threshold_result": threshold_result,
        "confidence_threshold": confidence_threshold,
        "note": "Threshold chỉ áp dụng selective prediction; không biểu thị accuracy tổng thể.",
    }, f, ensure_ascii=False, indent=2)

# Lưu state_dict để tương thích nếu backend nạp trọng số thủ công.
torch.save(model.state_dict(), os.path.join(OUTPUT_DIR, "phobert_vsmec_classifier.pt"))
# ============================================================
# 10. HÀM DỰ ĐOÁN CHO BACKEND: NHÃN + CONFIDENCE + V-A
# ============================================================
VA_ANCHORS = {
    0: [-0.8,  0.8],  # Anger
    1: [-0.6,  0.4],  # Disgust
    2: [ 0.8,  0.7],  # Enjoyment
    3: [-0.7,  0.9],  # Fear
    4: [ 0.0,  0.0],  # Other
    5: [-0.8, -0.6],  # Sadness
    6: [ 0.5,  0.9],  # Surprise
}


def predict_emotion_with_confidence(text, threshold=None):
    """Phân loại câu; accepted chỉ đánh dấu vượt threshold nếu có threshold."""
    if not isinstance(text, str) or not text.strip():
        raise ValueError("Văn bản dự đoán không được để trống.")

    processed = preprocess_vietnamese_text(text)
    encoded = tokenizer(
        processed,
        return_tensors="pt",
        max_length=MAX_LEN,
        padding="max_length",
        truncation=True,
    )
    encoded = {key: value.to(device) for key, value in encoded.items()}

    model.eval()
    with torch.no_grad():
        logits = model(**encoded).logits
        probabilities = torch.softmax(logits, dim=-1)[0].cpu().numpy()

    predicted_id = int(np.argmax(probabilities))
    confidence = float(probabilities[predicted_id])
    valence = float(sum(float(probabilities[i]) * VA_ANCHORS[i][0] for i in range(NUM_LABELS)))
    arousal = float(sum(float(probabilities[i]) * VA_ANCHORS[i][1] for i in range(NUM_LABELS)))

    return {
        "emotion": ID2LABEL[predicted_id],
        "confidence": round(confidence, 4),
        "accepted": None if threshold is None else confidence >= threshold,
        "valence": round(valence, 4),
        "arousal": round(arousal, 4),
        "probabilities": {
            ID2LABEL[i]: round(float(probabilities[i]), 4)
            for i in range(NUM_LABELS)
        },
    }


logger.info("===== TEST CÂU MẪU =====")
test_sentences = [
    "Phim này cuốn dã man, coi cười đau cả bụng 😂",
    "Chán quá, hôm nay toàn chuyện xui xẻo.",
    "Sợ quá đi mất, giật cả mình 😱",
]
for sentence in test_sentences:
    result = predict_emotion_with_confidence(sentence, confidence_threshold)
    logger.info("Câu: %s", sentence)
    logger.info("Kết quả: %s", json.dumps(result, ensure_ascii=False))

logger.info("Hoàn tất. Toàn bộ kết quả đã lưu tại: %s", OUTPUT_DIR)
print(f"\nHoàn tất. Kết quả được lưu tại: {OUTPUT_DIR}")
print(f"Model tốt nhất: {BEST_MODEL_DIR}")
print(f"Test Accuracy tổng thể: {test_result['accuracy']:.2%}")
print(f"Test Macro F1: {test_result['macro_f1']:.4f}")
print(
    "LƯU Ý: threshold (nếu tìm được) nhắm tới accuracy trên tập con được nhận; "
    "nó không nâng Test Accuracy tổng thể."
)
```
